Why are Passives excluded from NPS?

They are considered neutral customers. They affect the denominator (total responses) but not the NPS score directly.

NPS measures customer loyalty and satisfaction based on a single question

  - How likely are you to recommend our product/service to a friend or colleague
  - Customers rate from 0 to 10.

    Score	Category
    9-10	Promoters
    7-8	    Passives
    0-6	    Detractors

  - NPS = (% Promoters) - (% Detractors)

Can NPS be negative?  Yes 
  - Promoters = 20%
  - Detractors = 60%
  - NPS = -40

```python
#basic example
import numpy as np

scores = np.array([10, 9, 8, 7, 6, 5, 10, 9])
promoters = np.sum(scores>=9)
detractors = np.sum(scores<=6)
nps = (promoters/len(scores)*100)-(detractors/len(scores)*100)
print(f"nps is :{nps}")
```

```python
#reusable nps calculation
def calculate_nps(scores):
    promoters = np.sum(scores>=9)
    detractors = np.sum(scores<=6)
    return (promoters/len(scores)*100)-(detractors/len(scores)*100)

nps = calculate_nps(scores=scores)
print(f"nps is :{nps}")
```

```python
#additonal metrics examples

def additional_nps_metrics(scores):
    promoters = np.sum(scores >= 9)
    passives = np.sum((scores >= 7) & (scores <= 8))
    detractors = np.sum(scores <= 6)
    print("Promoters:", promoters)
    print("Passives:", passives)
    print("Detractors:", detractors)    

additional_nps_metrics(scores=scores)
```

```python
#advacned exampel - nps by region
data = [
    ("India",10),
    ("India",9),
    ("India",5),
    ("US",10),
    ("US",8),
    ("US",3)
]
from collections import defaultdict
import numpy as np

scores_by_region = defaultdict(list)
for region,score in data:
    scores_by_region[region].append(score)

for region,score in scores_by_region.items():
    arr = np.array(scores)
    promoters = np.sum(arr >= 9)
    detractors = np.sum(arr <= 6)

    nps = (
        promoters/len(arr)*100
        - detractors/len(arr)*100
    )

    print(region, nps)
```

```python
#vectorized NPS Calculation
import numpy as np

scores = np.random.randint(
    0,
    11,
    1_000_000
)

promoters_pct = np.mean(scores >= 9) * 100

detractors_pct = np.mean(scores <= 6) * 100

nps = promoters_pct - detractors_pct

print(nps)
```

```python
#monthly NPS trend- adavanced product analystics 
months = np.array([
    "Jan","Jan","Jan",
    "Feb","Feb","Feb"
])

scores = np.array([
    10,9,5,
    10,10,7
])

unique_months = np.unique(months)
for month in unique_months:
    month_scores = scores[months==month]
    promoters = np.mean(month_scores >= 9) * 100
    detractors = np.mean(month_scores <= 6) * 100
    nps = promoters - detractors
    print(month, nps)
```
