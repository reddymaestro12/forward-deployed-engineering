```python
import numpy as np
```

```python
# 1D Array
arr_1d = np.array([10, 20, 30, 40, 50])

print(f"max :{np.max(arr_1d)}")
print(f"min:{np.min(arr_1d)}")
print(f"sum:{np.sum(arr_1d)}")
print(f"average is:{np.mean(arr_1d)}")
print(f"std is :{np.std(arr_1d)}")
print(f"variance is:{np.var(arr_1d)}")
print(f"product is :{np.prod(arr_1d)}")
print(f"cummulative prod is :{np.cumulative_prod(arr_1d)}")
print(f"cumulative sum is :{np.cumulative_sum(arr_1d)}")
```

```python
# 2D Array
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
```

```python
#aggregations on entire array
print("Sum:", np.sum(arr_2d))
print("Mean:", np.mean(arr_2d))
print("Min:", np.min(arr_2d))
print("Max:", np.max(arr_2d))
```

```python
#aggregations on row-wise
print("Sum:", np.sum(arr_2d,axis=1))
print("Mean:", np.mean(arr_2d,axis=1))
print("Min:", np.min(arr_2d,axis=1))
print("Max:", np.max(arr_2d,axis=1))
```

```python
#aggregations on columns
print("Sum:", np.sum(arr_2d,axis=0))
print("Mean:", np.mean(arr_2d,axis=0))
print("Min:", np.min(arr_2d,axis=0))
print("Max:", np.max(arr_2d,axis=0))
```

```python
#find the indexex of the min,max elements
print(f"index of the {np.min(arr_1d)} is :{np.argmin(arr_1d)}")
print(f"index of the {np.max(arr_1d)} is :{np.argmax(arr_1d)}")
```

```python
import numpy as np
#array without duplicates
arr_1d = np.array([40, 10, 30, 50, 20])
#ascendig order
print(np.sort(arr_1d)) 
#descendig order - sort and reverse the array
print(np.sort(arr_1d)[::-1])

#array with duplicates
```

```python
#sorting on 2d array
arr_2d = np.array([
    [30, 10, 20],
    [60, 40, 50],
    [90, 70, 80]
])
```

```python
#row wise sorting
print(np.sort(arr_2d,axis=1))
```

```python
#column wise sorting 
print(np.sort(arr_2d,axis=0))
```

```python
#sorting based on a specific column
employees = np.array([
    [101, 28, 50000],
    [102, 22, 45000],
    [103, 35, 70000],
    [104, 30, 60000]
])
```

```python
#sort the dataset by age which is second colum
sortby_age=employees[
employees[:, 1].argsort()
]
print(sortby_age)
```

```python
#sort by salary descending
print(employees[employees[:,2].argsort()])
```
