```python
#without vectorization iterations are required
a = [1, 2, 3, 4]
b = [10, 20, 30, 40]

result = []
for i in range(len(a)):
    result.append(a[i] + b[i])

print(result)
```

```python
def even_odd(a):
    if a%2==0:
        a=a+2
    else:
        a=a-1
    return a
#with vectorization
import numpy as np
arr = np.array([1,2,3,4,5,6])
f = np.vectorize(even_odd)
print(f(arr))

numbers = np.array([1, 2, 3, 4, 5])
squared = numbers ** 2
print(squared)
```

```python
#conditional operations - get numbmers >5
numbers = np.array([2, 4, 6, 8, 10])
print(numbers[numbers>5])
```

```python
#data normalization data pre-processing step 
data = np.array([10, 20, 30, 40, 50])
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print(normalized)
```

```python
#Vectorization in Machine Learning - computing predictions
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

weights = np.array([0.5, 1.5])
bias = 2
predictions = X @ weights + bias

print(predictions)
```

```python
#broadcast the second array "array2" to match the shape of the first array "array1" 
#internally np.tile() works and the smaller array shape is reshaped to the bigger array with same data elememnts
#Creating two arrays with different shapes
array1 = np.array([[1, 2, 3], [4, 5, 6]])     
array2 = np.array([10, 20, 30])              

# Adding arrays with broadcasting
result = array1 + array2
print(result)
```

#broadcasting rules 
- same shape : broad casting is allowed
- left array should be column matrix and right array should be row matrix :BC is allowed
- columns in first matrix and rows in second matrix should match : bc is allwoed
