- Maksing is a process to filter,select or modify the elements in an array based on a condition.
- Mask is a bollean array (True/False) of same shape of original array
- best advanatge is filtering can be done without looping
- values can be replaced with masking

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
mask = arr >25
print(mask) #import numpy as np

#if we want to see the values associated with the arrray
print(arr[arr>25])
```

```python
arr = np.arange(20)
even = arr%2==0
print(even)

print(arr[arr%2==0])
```

```python
#array valies can be repalced with masking
r =  np.array([10,-1,-3,-5,4])
print(r)
#replace valleues <0 wiht 100
r[r<0]=100
print(r)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])
#get values between 20 and 50 - multiple conditions are allowed
mask = (arr >= 20) & (arr <= 50)

print(arr[mask])

mask1 = (arr < 20) | (arr > 40)

print(arr[mask1])
```

Masking on 2D Arrays

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

mask = arr > 50
print(mask)
#output becomes a flattened 1D array.
print(arr[mask])
```

```python
#useful in data cleaning - remove missing values represented by 999
data = np.array([10, 20, -999, 40, -999, 50])
cleaned = data[data!=-999]
print(cleaned)
```

```python
#conditioned based values
prices = np.array([10, 20, 30, 40, 50])
budget=30
any_test = np.any(prices==budget)
print(any_test)

all_test=np.any(prices==budget)
print(all_test)

where_test = np.where(prices>25,"high","low")
print(where_test)
```
