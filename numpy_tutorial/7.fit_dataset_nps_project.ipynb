```python
import numpy as np
```

```python
data = np.loadtxt(
    r"E:\fde\numpy_tutorial\fit.csv",
    delimiter=",",
    dtype=str,
    skiprows=1
)
```

```python
# Extract features
step_count = data[:, 1].astype(int)
mood = data[:, 2]
calories = data[:, 3].astype(int)
sleep = data[:, 4].astype(float)
activity = data[:, 5]
```

```python
#average stepcount on Active days
avg_active_stepcount = np.mean(step_count[data[:, 5] == 'Active'])

print("Average active step count:", avg_active_stepcount)
```

```python
sad_percentage = np.mean(mood == 'Sad') * 100

print(f"Percentage of days with Sad mood: {sad_percentage:.2f}%")
```

```python
#how many days had >500 steps and burned 500 calories
print(np.sum((step_count>500) & (calories==500)))
```
